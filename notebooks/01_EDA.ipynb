# Key Business Insights

After performing Exploratory Data Analysis (EDA), the following key observations were identified:

- Around **26.5%** of customers have churned, making this a moderately imbalanced classification problem.
- Customers with **month-to-month contracts** are significantly more likely to churn compared to customers on yearly contracts.
- Customers with **shorter tenure** are more likely to leave the company.
- **Fiber Optic** users exhibit a higher churn rate than DSL users.
- Customers paying through **Electronic Check** show the highest churn among all payment methods.
- Customers without **Online Security** are much more likely to churn.
- Customers without **Tech Support** have considerably higher churn.
- Customers without **Partners** show higher churn compared to customers with partners.
- Customers without **Dependents** are more likely to churn.
- Numerical features show only weak correlations, indicating low multicollinearity among variables.

These findings provide valuable guidance for feature engineering and model development in the next phase.

# EDA Summary

The exploratory data analysis provided a comprehensive understanding of the customer churn dataset.

Major findings indicate that contract type, tenure, internet service, payment method, online security, tech support, partner status, and dependents have strong relationships with customer churn.

The dataset contains both numerical and categorical variables with minimal multicollinearity. No major data quality issues were observed apart from preprocessing requirements such as encoding categorical variables and scaling numerical features where appropriate.

The insights gained during EDA will guide the feature engineering and model-building stages of this project.
